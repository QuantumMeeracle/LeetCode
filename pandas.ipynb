```python
#ignore index creates the right order for the records when combined


india_weather = pd.DataFrame({
    'city': ['mumbai', 'delhi', 'bangalore'],
    'temperature': [32, 45, 30],
    'humidity': [80, 20, 78]

})


us_weather = pd.DataFrame({
    'city': ['new york', 'chicago', 'orlando'],
    'temperature': [21, 14, 35],
    'humidity': [68, 65, 75]
})


df = pd.concat([india_weather, us_weather], ignore_index=True)
df
```

```python
#row-wise concatenation

df = pd.concat([india_weather, us_weather], axis=1)
df
```

```python
#column-wise concatenation (default)

df = pd.concat([india_weather, us_weather], axis=0, ignore_index=True)
df
```

```python
#combining dataframes using merge function (outer, inner, left, right joins)

df = pd.merge(india_weather, us_weather, on='city', how='right')
df
```
