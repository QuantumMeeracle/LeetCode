```python
s = {1, 2, 3, 4, 5}

s2 = {4, 5, 6, 7, 8}

```

```python
s.union(s2)  # Returns a new set with elements from both sets
s.intersection(s2)  # Returns a new set with elements common to both sets
s.difference(s2)  # Returns a new set with elements in s but not in s2
s.symmetric_difference(s2)  # Returns a new set with elements in either s or s2 but not in both

```

```python
s.issubset(s2)  # Checks if s2 is a subset of s
```

```python
def sqrite_map(n):
    return n*n

def filter_even(n):
    if n%2 == 0 and n>0:
        return n

lst = list(range(-5,5))

list(map(sqrite_map, lst))

```

```python
lst = list(range(-5,5))
list(filter(filter_even, lst))
```

```python
from functools import reduce

def add(x, y):
    return x + y

reduce(add, lst)  # (((-5 + -4) + -3) + ... + 4)
```
