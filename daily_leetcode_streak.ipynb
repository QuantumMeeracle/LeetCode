# Best solution for time complexity (Pascal triangle problem)



```python
# updated solution best time complexity O(n)

from math import comb
from typing import List

class Solution:
    def triangularSum(self, nums: List[int]) -> int:
        lnums = len(nums)

        result = 0

        for i,j in enumerate(nums):
            result = result + (j*comb(lnums-1,i))

        return result % 10
```

```python
# My answer:
class Solution:
    def numWaterBottles(self, numBottles: int, numExchange: int) -> int:
         drkup = numBottles

         while True:
            rem_count = numBottles % numExchange
            extra_bottles = (numBottles // numExchange)

            if not extra_bottles < 1:
                drkup = drkup + extra_bottles
                numBottles = extra_bottles + rem_count
            else:
                break

         return drkup
```
